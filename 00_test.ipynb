```python
# Student grade A - Above 95 marks, B - 80 to 95 marks, C - 70 to 80 marks, D - 60 to 70 marks, E - 50 to 60 marks, F - Below 50 marks
# If the student has scored more than 95 marks, then print "A", if the student has scored between 80 to 95 marks, then print "B", if the student has scored between 70 to 80 marks, then print "C", if the student has scored between 60 to 70 marks, then print "D", if the student has scored between 50 to 60 marks, then print "E", if the student has scored below 50 marks, then print "F".
marks = int(input("Enter your marks: "))
if marks > 95:
    print("A")
elif marks >= 80 and marks <= 95:
    print("B")
elif marks >= 70 and marks < 80:
    print("C")
elif marks >= 60 and marks < 70:
    print("D")
elif marks >= 50 and marks < 60:
    print("E")
else:
    print("F")

```
